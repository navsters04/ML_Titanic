```python
#import library
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
#import dataset
df = pd.read_csv('data/titanic.csv')
df.head()
```

From the data, we can remove the "PassengerId" "Name" and "Ticket" column, because it represent unique identifiers that do not carry generalizable signal for predicting survival.
We can also remove "Cabin" because it has 687 missing rows out of 891 (77%), which is too much to guess accurately

```python
# Delete unrelevant columns
df = df.drop(columns=['PassengerId', 'Name', 'Ticket', 'Cabin'],errors='ignore')
```

To fill in the missing data, We will the median age (28.0) instead of the mean (29.7) because age is right-skewed with extreme values, making the median a safer choice against outliers.
For Embarked, we will mode imputation because it is a categorical feature missing only 2 out of 891 entries, so filling them with the most common port ('S') maintains the data's original distribution.

```python
#Fill null values
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df['Age'] = df['Age'].fillna(df['Age'].median())

df.isnull().sum()
```

```python
#Change sex to 0 and 1
df['Sex'] = df['Sex'].replace({'male': 0, 'female': 1})
```

```python
#Change embarked to 0 and 1
df = pd.get_dummies(df, columns=['Embarked'], drop_first=True)
df['Embarked_Q'] = df['Embarked_Q'].astype(int)
df['Embarked_S'] = df['Embarked_S'].astype(int)
```

Numerical features like "Age" (0–80) and "Fare" (0–512) operate on different scales. Standardizing them ensures linear models do not get dominated by the larger values.

```python
scaler = StandardScaler()
# Standardize numerical features
df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])
```

```python
# Target (y)
y = df['Survived']

# Fitur (X)
x = df.drop(columns=['Survived'])

x.shape,y.shape
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42, stratify=y)

x_train.shape,x_test.shape
```
