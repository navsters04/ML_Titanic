```python
#import library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#import dataset
df = pd.read_csv('data/titanic.csv')
df.head()
```

```python
# Show statistics
df.describe()
# df.describe(include=['object','string'])
```

```python
# Show null values
df.isnull().sum()
```

From the data above, we can see that there is 177 age data, 687 cabin data, and 2 embarked data missing

```python
#Show the survival count of the passengers
sns.countplot(df, x='Survived')
plt.title('Survival count')
plt.show()
```

```python
#Show the classes of the passengers
sns.countplot(df, x='Pclass')
plt.title('Passenger Classes')
plt.show()
```

```python
#Show the age of the passengers
sns.histplot(df, x='Age')
plt.title('Passenger Age')
plt.show()
```

```python
#Show the gender of the passengers
sns.countplot(data=df, x='Sex')
plt.title('Passenger Genders')
plt.show()
```

## Relationship Between Features

```python
sns.barplot(df, x='Sex', y='Survived', hue='Pclass')
plt.show()
```

```python
df.groupby('Sex')['Survived'].mean()
```

```python
df.groupby('Pclass')['Survived'].mean()
```

```python
sns.kdeplot(data=df, x='Age', hue='Survived')
plt.title('Age Distribution by Survival')
plt.show()
```

### Conclusion ###

1. Female passengers had a significantly higher survival rate (74%) compared to male passengers (19%).
2. First-class passengers had the highest survival rate (63%), followed by second-class (47%), and third-class with the lowest (24%).
3. The Age, Cabin, and Embarked columns need to be handled during preprocessing as they contain null values.
